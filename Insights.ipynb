```python
import json
from deep_translator import GoogleTranslator

# 1. Load the original Chinese file
with open("XES3G5M/metadata/kc_routes_map.json", "r", encoding="utf-8") as file:
    kc_map = json.load(file)

# 2. Set up the translator and a new dictionary for the English data
translator = GoogleTranslator(source='zh-CN', target='en')
english_kc_map = {}

print("Translating 1175 items... this will take a minute or two.")

# 3. Loop through and translate each concept
for kc_id, chinese_text in kc_map.items():
    try:
        english_text = translator.translate(chinese_text)
        english_kc_map[kc_id] = english_text
    except Exception as e:
        # If one fails, just print an error and move to the next
        print(f"Failed to translate ID {kc_id}")
        english_kc_map[kc_id] = chinese_text 

# 4. Save the brand new English file
with open("XES3G5M/metadata/kc_routes_map_en.json", "w", encoding="utf-8") as file:
    json.dump(english_kc_map, file, indent=4)

print("Done! Your new English file is saved as: kc_routes_map_en.json")
```
